## Logical Conditions :


*   Equals: ```a==b```
*   Not Equals: ```a != b```
*   Less than: ```a < b ```
* Less than or equal to: ``` a <= b```
* Greater than: ```a > b```
* Greater than or equal to: ```a >= b```

---




## IF condition
> If condition is used to execute a block of code only when a specified condition evaluates to True

````python
a = 33
b = 200

if b > a:
  print(" b is greater than a")
````

````python
# Multiple Statements in If Block
age = 20
if age >= 18:
  print("You are an adult")
  print("You can vote")
  print("You have full legal rights")
````

````python
# Boolean variables can be used directly

is_logged_in = True
if is_logged_in:
  print("Welcome Back!")
````

## IF-ELSE Statement
> if else statment is used to execute one block of code when  the condition is True and anothe block when the confition is False.

````python
age = 10
if age <= 12:
  print("Travel for free.")
else:
  print("Pay for ticket.")
````

## if-elif-else Statement
> Used to check multiple conditions in a program. It executes a block of code when its condition evaluates to True after previous condition evaluate to False.

````python
age = 25

if age <= 12:
  print("Chile.")
elif age <= 19:
  print("Teenager")
elif age <= 35:
  print("Young Adult")
elif age <= 55:
  print("Adult")
else:
  print("Senior Citizen")
````

## Nested If-else Statement
> if-else statement placed inside another f or else block. it is used to check conditions within another condition.

````python
age = int(input("Enter your age: "))
is_member = True

if age >= 60:
  if is_member:
    print("30% senior discount!")
  else:
    print("20% senior discount!")
else:
  print("No discount!")
````

## Conditional Expression (Ternary Operator)
> Short way to write an if-else statement in a single line. It is used when choosing between two values based on a condition.

````python
age = 20

s = "Adult" if age >= 18 else "Minor"

print(s)
````

## Match-Case Statement
> Used to compare a value against multiple patterns and execute the matching blaock of code. similar to switch-case statement in other programming languages.

````python
number = 2

match number:
  case 1:
    print("One")
  case 2 | 3:
    print("Two or Three")
  case _:
    print("Other number")
````
