## Loops
> Loops are used to execute a block of code repetedly until a confition is met or all items in a sequence are procesed.

## For Loop
> Used to iterate over a sequence such as a list, tuple, string or range. It executes a block of code once for each item in the sequence.

```python
n = 4

for i in range(0,n):
  print(i)
```

### Iterating by index of Sequences

```python
a = ["Python", "for", "Machine", "Learning"]

for idx in range(len(a)):
  print(a[idx])
```

## While Loop
> repetedly executes a block of code as long as the given condition remain is true. When the condition becomes false, the line immediately after the loop in the program is executed.

```python
count = 0
while (count<3):
  count = count + 1
  print("I am Batman")
```

## Nested Loops
> A nested loop is a loop inside another loop. The inner loop executes completely for every iteration of the outer loop

```python
for i in range(1, 5):
  for j in range(i):
    print(i, end=' ')
  print()
```
