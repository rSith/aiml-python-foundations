# Functions
> Functions aare reusable blocks of code used to perform a specific task.

A function can be defined using `def` keyword.


```

def function_name(parameters)
    statement
  return expression
```



````python
#defining a function
def hello():
  print("Hello, World!")

#Calling a function
hello()
````

## Function Arguments
> Arguments are vbalues passed to a fucntion when it is called.


```
def function_name(arguments):
    function body
  return value
```



````python
def evenOdd(x):
  if (x % 2 == 0):
    return "Even"
  else:
    return "Odd"

print(evenOdd(16))
print(evenOdd(17))
````

## Types of Function Arguments
01. **Default Argument**: Uses a predefined value when no value is passed during the function call.
02. **Keyword Argument**: pass values using parameter names, so argument order does not matter.
03.  **Positional Argument**: Values are assigned to parameters based on thire order in the function call.
04. **Arbitrary Arguments**: Allow functions to accept multiple values.Two Special symbols:
    - `*args` collects extra positional arguments as a tuple.
    - `**kwargs` collects extra keyword arguments as a dictionary.


````python
#Default Argument
def my_function(name = "friend"):
  print("Hello", name)

my_function("Kamal")
my_function("Nimal")
my_function()
````

````python
# Keyword Argument
def my_function(animal, name):
  print("I have a", animal)
  print("My",animal + "'s name is",name)

my_function(animal = "dog", name = "Bruno")
````

````python
# Positional Arguments
def my_function(animal, name):
  print("I have a", animal)
  print("My",animal + "'s name is",name)

my_function("dog","Bruno")
````

````python
# Arbitrary Arguments - *args
# it allows a fucntion to accept any numbver of positional arguments.

def my_function(*kids):
  print("The youngest child is " + kids[2])

def my_function2(greeting, *names):
  for name in names:
    print(greeting + ", " + name)

my_function("Emil", "Tobias", "Linus")
my_function2("Hello","Emil", "Tobias", "Linus")
````

````python
# Arbitrary Arguments - **kwargs
# Allows a function to accept any number of keyword arguments.

def display_profile(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

display_profile(name="Alice", age=30, job="Engineer")
````
