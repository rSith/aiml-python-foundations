# Python Collections(Arrays)
There are four collection data types in the Python programming langugae.
- **Lists**: is a collection which is ordered and changeable. Allows duplicate members.
- **Tuple**: is a collection which is ordered and unchangeable. Allows duplicate members.
- **Set**: is a collection which is unorderd,unchangeable*(but you can remove and add items whenever you like)*, and unindexed. No duplicate members.
- **Dictionary**: is a collection which is orderd and changeable. No duplicate members.

## 01. Storing Data in Lists
> A list is an ordered, mutable collection of items enclosed in square brackes. Lists can store any data type ( integers, strings, floats, booleans, ) or mixed types.

- It is also possible to se the `list()` constructor when creating a new list.

````python
fruits = ['apple', 'banana', 'cherry']
mixed_list = [1, 'hello', 3.14, True]
nested_list = [[1, 2, 3], ['a', 'b', 'c']]

print(fruits)
print(mixed_list)
print(nested_list)
````

````python
fruits = list(('apple', 'banana', 'cherry'))
print(fruits)
````

##02. Accessing List Data
> Items in a list are accessed using zero-based indexing


*   Positive indexing starts from 0 for the first element.
*   Negative indexing starts from -1 for the last element.
*   Slicing allows extracting a sublist using list **[start: stop: step]**



````python
first_item = fruits[0]
last_item = fruits[-1]
sublist = mixed_list[1:3:1]
nested_item = nested_list[0][1]

print(first_item)
print(last_item)
print(sublist)
print(nested_item)
````

### Loop Through a List using  for loop:

````python
# Loop through the list items by using a for loop:
fruits = ['apple', 'banana', 'cherry']

for i in fruits:
  print(i)
````

````python
# Loop Through the Index Numbers

fruits = ['apple', 'banana', 'cherry']

for i in range(len(fruits)):
  print(fruits[i])
````

### Loop Through a List using while loop:

````python
fruits = ['apple', 'banana', 'cherry']

i = 0;
while i < len(fruits):
  print(fruits[i])
  i = i + 1
````

## 03. Change Item Value
- To change the value of a specific item, refer to the index number.
- To change the value of items within a specific range, define a list with the new values, and refer ti the range of index numbers where you want to insert the new values.

````python
fruits = ['apple', 'banana', 'cherry']
fruits[1] = 'blueberry'
print(fruits)

fruits[1:3] = ['kiwi', 'orange']
print(fruits)
````

## 04. Manipulating List Data
> Lists are mutable, meaning their contents can be changed after creation.


1.   **Adding items:**
  - ```.append()``` adds to the end,
  - ```.insert()``` inserts at a specific index
2. **Extend List:**
  - To append elements from another list to the current list, use the `extend()` method.
3. **Removing items:**
  - ```.remove()``` delete by the value,
  - ```.pop()``` removes and returns an item.
  - ```.del()``` deletes at an index
  - `.clear()` empties the list. the list still remains, but it has no content.
4. **Modifying items:**
  - ```fruits[0] = 'mango'``` replace element at index 0.
5. **Utility methods:**
  - ```.len()``` returns length,
  - ```.sort()``` sort the list alphanumerically, ascending.
  - `.sort(reverse = True)` sort desending
  - ```reverse()``` reverses in-place.

6. **Copy a List:**
- `.copy()` or `.list()` to copy a list.
7. **Join Two Lists:**
- use `+` operator
- use `.extend()` method
