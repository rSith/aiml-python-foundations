## 01. Variables
A variable is a name that refers to a value.

**Creating Variables:**
> use an assignment statement (the = operator) to create a new variables and assign values to them





```python
#Assigning a string to a variable
message = "Hello World!"

#Assigning an integer to a variable
n = 17
```

## 02. Data Types
> Every value belongs to a specific "type". The type of a variable is simply the type of the value it currently refers to.


### Strings (str)
Sequence of characters.

```python
greeting = "Hello, World!"
```

### Integers (int)
Represent whole numbers. Can perform standard mathematical operations on them.

```python
age = 25
score = 100
```

### Floating-Point Numbers (float)
Any number with a decimal point.

```python
pi = 3.14159
temperature = 98.6
```

### Booleans (bool)
Only two special values: True or False. These are often the result of conditional tests or boolean expressions.

```python
is_active = True
is_game_over = False

# Boolean are often generated by comparisons
result = (5 == 5) #Evaluates to True
```

### Checking Data Types
use built-in `type()` function

```python
print(type(greeting))
print(type(age))
print(type(pi))
print(type(is_active))
```
